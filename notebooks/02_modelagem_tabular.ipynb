```python
# Celula 1 - Imports
import sys, os
from pathlib import Path

PROJECT_ROOT = Path.cwd().resolve()
if not (PROJECT_ROOT / "src").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.insert(0, str(PROJECT_ROOT))
os.chdir(PROJECT_ROOT)

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.cancer.train import train_cancer_models
from src.cancer.predict import CancerPredictor

sns.set_theme(style='whitegrid')
print('Imports carregados')
```

```python
# Celula 2 - Treinar modelos (salva em models/cancer/)
modelos, resultados, scaler, features = train_cancer_models()
print('Treino concluido!')
```

```python
# Celula 3 - Tabela de metricas por modelo
df_resultados = pd.DataFrame(resultados).T
df_resultados = df_resultados.round(4) * 100
df_resultados.columns = ['Acuracia', 'AUC', 'Recall', 'Precisao', 'F1']
df_resultados.index.name = 'Modelo'
print('Metricas (%) por modelo:')
df_resultados
```

```python
# Celula 4 - Grafico comparativo
df_resultados.plot(kind='bar', figsize=(12, 6))
plt.title('Comparativo de Metricas - Cancer de Mama', fontsize=14, fontweight='bold')
plt.ylabel('Score (%)')
plt.xlabel('Modelo')
plt.ylim(50, 100)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# Tambem salvo em outputs/cancer/comparativo.png
import os
os.makedirs('outputs/cancer', exist_ok=True)
plt.savefig('outputs/cancer/comparativo_nb.png', dpi=150, bbox_inches='tight')
print('Grafico salvo em outputs/cancer/comparativo_nb.png')
```

```python
# Celula 5 - Melhor modelo e predicao de exemplo
melhor = df_resultados['F1'].idxmax()
print(f'Melhor modelo por F1: {melhor} ({df_resultados.loc[melhor, "F1"]:.2f}%)')

# Exemplo: primeira amostra do dataset (M - maligno)
predictor = CancerPredictor()
amostra = [17.99, 10.38, 122.8, 1001.0, 0.1184, 0.2776, 0.3001,
           0.1471, 0.2419, 0.07871, 1.095, 0.9053, 8.589, 153.4,
           0.006399, 0.04904, 0.05373, 0.01587, 0.03003, 0.006193,
           25.38, 17.33, 184.6, 2019.0, 0.1622, 0.6656, 0.7119,
           0.2654, 0.4601, 0.1189]
result = predictor.predict(amostra, melhor)
print('\nPredicao de exemplo:')
for k, v in result.items():
    if k != 'features':
        print(f'  {k}: {v}')
```

```python
# Celula 6 - Feature importance (Random Forest)
rf = modelos['random_forest']
importances = rf.feature_importances_
indices = np.argsort(importances)[::-1]

plt.figure(figsize=(10, 8))
plt.barh(range(15), importances[indices[:15]][::-1], color='steelblue')
plt.yticks(range(15), [features[i] for i in indices[:15]][::-1])
plt.title('Feature Importance - Random Forest (Top 15)', fontsize=14, fontweight='bold')
plt.xlabel('Importancia')
plt.tight_layout()
plt.show()

print('Top 10 features mais importantes:')
for i in range(10):
    print(f'  {i+1}. {features[indices[i]]}: {importances[indices[i]]:.4f}')
```

# Preparação das mamografias CBIS-DDSM

Cria imagens JPG rotuladas em `data/images/cbis-ddsm/` a partir do ZIP fornecido.

```python
import sys, os
from pathlib import Path

PROJECT_ROOT = Path.cwd().resolve()
if not (PROJECT_ROOT / 'src').exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.insert(0, str(PROJECT_ROOT))
os.chdir(PROJECT_ROOT)
```

```python
from pathlib import Path
from src.cnn.dataset import (
    CBIS_BASE_DIR,
    CBIS_EXTRACTED_PATH,
    CBIS_SPLITS,
    CBIS_CLASSES,
    prepare_cbis_ddsm_dataset,
    prepare_extracted_cbis_ddsm_dataset,
)


def _contagem_por_diretorio(base: Path) -> dict:
    """Conta JPGs/PNGs por split/classe já preparados."""
    contagem = {}
    for split in CBIS_SPLITS:
        for label in CBIS_CLASSES:
            d = base / split / label
            contagem[f"{split}_{label}"] = (
                len(list(d.glob("*.jpg"))) + len(list(d.glob("*.png"))) if d.exists() else 0
            )
    return contagem


summary = _contagem_por_diretorio(CBIS_BASE_DIR)
if CBIS_BASE_DIR.exists() and all(v > 0 for v in summary.values()):
    print("Dataset já preparado em data/images/cbis-ddsm — nada a fazer.")
elif CBIS_EXTRACTED_PATH.exists():
    print("Preparando splits a partir da extração local (jpeg + csv)...")
    summary = prepare_extracted_cbis_ddsm_dataset()
else:
    print("Preparando splits a partir do ZIP CBIS-DDSM...")
    summary = prepare_cbis_ddsm_dataset()

print("\nResumo da preparação:")
for k, v in summary.items():
    print(f"  {k}: {v}")
summary
```
